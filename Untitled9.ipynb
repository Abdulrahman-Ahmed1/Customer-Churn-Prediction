```python
x = df[['Age', 'Gender', 'Tenure', 'MonthlyCharges']]
y = df[['Churn']]
```

```python
x['Gender']
```

```python
y['Churn']
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
y['Churn'] = LabelEncoder().fit_transform(y['Churn'])
# Yes = 1 ----------- No = 0
```

```python
y['Churn'].head(10)
```

```python
x['Gender'] = LabelEncoder().fit_transform(x['Gender'])
# Male = 1 --------- Female = 0
```

```python
x['Gender'].head(10)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
import joblib
joblib.dump(scaler, 'scaler.pkl')
```

```python
from sklearn.metrics import accuracy_score
def modelperformance(predictions):
  print('Accuracy Score: ', accuracy_score(y_test, predictions))
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.linear_model import LogisticRegression
log_model = LogisticRegression()
log_model.fit(X_train, y_train)
```

```python
y_pred = log_model.predict(X_test)
y_pred
```

```python
modelperformance(y_pred)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model = KNeighborsClassifier()
knn_model.fit(X_train, y_train)
```

```python
y_pred = knn_model.predict(X_test)
y_pred
```

```python
modelperformance(y_pred)
```

```python
from sklearn.svm import SVC
svm = SVC()
```

```python
param_gird = {'C': [0.01, 0.1, 0.5, 1],
              'kernel': ['linear', 'poly', 'rbf', 'sigmoid']
             }
```

```python
gridsvc = GridSearchCV(svm, param_gird, cv=5)
```

```python
gridsvc.fit(X_train, y_train)
```

```python
gridsvc.best_params_
```

```python
y_pred = gridsvc.predict(X_test)
y_pred
```

```python
modelperformance(y_pred)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
param_gird = {
    'criterion': ['gini', 'entropy'],
    'splitter': ['best', 'random'],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}
```

```python
grid_tree = GridSearchCV(DecisionTreeClassifier(), param_gird, cv=5)
```

```python
grid_tree.fit(X_train, y_train)
```

```python
y_pred = grid_tree.predict(X_test)
y_pred
```

```python
modelperformance(y_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc_model = RandomForestClassifier()
```

```python
param_gird = {
    'n_estimators': [32, 64, 128, 256],
    'max_features': [2, 3, 4],
    'bootstrap': [True, False]
}
```

```python
grid_rfc = GridSearchCV(rfc_model, param_gird, cv=5)
```

```python
grid_rfc.fit(X_train, y_train)
```

```python
grid_rfc.best_params_
```

```python
y_pred = grid_rfc.predict(X_test)
y_pred
```

```python
modelperformance(y_pred)
```

```python
best_model = grid_rfc.best_estimator_
```

```python
joblib.dump(best_model, 'best_model.pkl')
```
